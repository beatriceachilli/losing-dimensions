## 4. Check Notebook Compatibility

The existing visualization notebook indexes `singular_values_orthg` at 100 time steps. This quick check confirms that the pickle has the expected keys and length.


````python
import pickle
from pathlib import Path

pickle_path = Path('pickles/linear_model_spectrum_ds_20000_N_30_s_1_d1_5_d2_10_v1_normalized.pickle')
with pickle_path.open('rb') as handle:
    res = pickle.load(handle)

print(res.keys())
print(len(res['time_steps']))
print(len(res['singular_values_orthg']))
print(res['singular_values_orthg'][0].shape)
````
